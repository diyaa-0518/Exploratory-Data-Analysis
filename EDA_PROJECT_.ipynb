```python
#Create a bar chart of movies and TV shows.
df['type'].value_counts().plot(
    kind='bar',title='Movies and TV Shows'
)
plt.xlabel('Content Type')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Create a pie chart of content type.
df['type'].value_counts().plot(
    kind='pie', autopct= "%1.1f%%",
    title="Netflix Content Distribution"

)
plt.show()
```

```python
#Create a histogram of release years.
plt.hist(df["release_year"], bins=20)
plt.title("Distribution of Release Year")
plt.xlabel("Release Year")
plt.show()
```

```python
#Create a line chart of titles added each year.
df['date_added'].value_counts().sort_index().plot(
    kind='line', marker="*",
    title="Netflix Titles added by year"
)
plt.xlabel("Year Added")

plt.grid(True)
plt.show()
```

```python
#Create a box plot of movie durations.
duration=df.loc[
    df["type"]=="Movie","duration"
].dropna()

sns.boxplot(x=duration)
plt.title("Distribution Of Movie Duration")
plt.xlabel("Duration in minutes")
plt.show()
```

```python
#Create a heatmap of content type and rating.
rating=pd.crosstab(df["rating"],df["type"])
sns.heatmap(rating,annot=True,fmt="d",cmap="Blues")
plt.title("Content Rating By Type")
plt.show()
```

```python
#Convert date_added to datetime.
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
#create a new column Day Name and Month name from date_added column.
df['Month_Name']=df['date_added'].dt.month_name()
df["Day_Name"]=df['date_added'].dt.day_name()
df["Year"]=df['date_added'].dt.year.astype('Int64')
df

```
